#### Morphological Operations

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

## Erosion

```python
img = cv2.imread('j.png',0)
kernel = np.ones((5,5),np.uint8)
erosion1 = cv2.erode(img,kernel,iterations = 1)
erosion2 = cv2.erode(img,kernel,iterations = 2)
erosion3 = cv2.erode(img,kernel,iterations = 3)
```

```python
cv2.imshow('img',img)
cv2.imshow('dst1',erosion1)
cv2.imshow('dst2',erosion2)
cv2.imshow('dst3',erosion3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dilation1 = cv2.dilate(img,kernel,iterations = 1)
dilation2 = cv2.dilate(img,kernel,iterations = 2)
dilation3 = cv2.dilate(img,kernel,iterations = 3)
```

```python
cv2.imshow('img',img)
cv2.imshow('dst4',dilation1)
cv2.imshow('dst5',dilation2)
cv2.imshow('dst6',dilation3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
